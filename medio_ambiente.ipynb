```python
import pandas as pd
import requests # requests es una librería de Python que permite realizar peticiones HTTP a páginas web y servidores.
from charset_normalizer import from_bytes

# FASE DE CARGA Y EXPLORACIÓN
url = "https://www.juntadeandalucia.es/datosabiertos/portal/dataset/6b186c12-2fd9-450d-a325-02052e2fe352/resource/776bc77d-4680-4745-aa73-9a4b16beaa90/download/banderas_azules_2026-27.csv"

# Descargamos el archivo csv
respuesta = requests.get(url)
contenido = respuesta.content

# Detectar la codificación
codificacion = from_bytes(contenido).best()
print("Codificación detectada:", codificacion.encoding)

# Leemos el CSV aplicando el separador y la codificación correctos
df = pd.read_csv(
    url,
    sep=";",
    decimal=",",
    encoding= codificacion.encoding
)

# Anotamos filas, columnas, tipos y nulos según pide en la ficha
print()
print(df.head())
print()
print(df.shape)
print()
print(df.dtypes)
print()
print(df.isna().sum())
print()
print(df.describe())




```

```python

# FASE DE TRANSFORAMCIÓN
# Transformación 1: Limpiar nombres de columnas (elimina espacios inecesarios)
df.columns = df.columns.str.strip()

# Transformación 2: Eliminar registros duplicados
df = df.drop_duplicates()

# Transformación 3: Limpieza de datos, eliminar columnas innecesarias/nulas
df = df.drop(columns=['Notas'], errors='ignore')

## Transformación 4: Corregir el nombre de la columna AÑO
df = df.rename(columns={'Ańo': 'Año'})

# Transformación 5: Convertir tipos y estructurar datos
# Separamos la columna Año en dos columnas, rellenamos huecos y convertimos a entero
df[['Año inicio', 'Año final']] = df['Año'].str.split('-', expand=True)
df['Año final'] = df['Año final'].fillna(df['Año inicio'])

df['Año inicio'] = df['Año inicio'].astype(int)
df['Año final'] = df['Año final'].astype(int)
del df['Año']

print(df.dtypes)
print()

# Transformación 6: Ordenación del dataframe
# Agrupamos visualmente ordenando primero por Provincia y luego por Año de inicio (descendente)
df = df.sort_values(by=['Provincia', 'Año inicio'], ascending=[True, False])
df.head(10)
```
